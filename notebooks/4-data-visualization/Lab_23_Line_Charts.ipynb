### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
# ---- Helpers used throughout this lab ----
import numpy as np

# Age is stored as text ranges, so map each range to an approximate number (its midpoint)
AGE_MAP = {
    'Under 18 years old': 17, '18-24 years old': 21, '25-34 years old': 29.5,
    '35-44 years old': 39.5, '45-54 years old': 49.5, '55-64 years old': 59.5,
    '65 years or older': 70,
}
AGE_ORDER = list(AGE_MAP.keys())

# TimeSearching / TimeAnswering are text ranges -> approximate minutes per day
TIME_MAP = {
    'Less than 15 minutes a day': 7.5, '15-30 minutes a day': 22.5,
    '30-60 minutes a day': 45, '60-120 minutes a day': 90, 'Over 120 minutes a day': 150,
}

def years_to_num(series):
    """YearsCodePro is text: turn 'Less than 1 year' / 'More than 50 years' into numbers."""
    return pd.to_numeric(series.replace({'Less than 1 year': '0', 'More than 50 years': '51'}),
                         errors='coerce')

def split_counts(series, top_n=None):
    """Count items in a ';'-separated multi-answer column."""
    counts = series.dropna().str.split(';').explode().str.strip().value_counts()
    return counts.head(top_n) if top_n else counts

for col in ['ConvertedCompYearly', 'JobSatPoints_6', 'WorkExp']:
    df[col] = pd.to_numeric(df[col], errors='coerce')
df['YearsCodePro_num'] = years_to_num(df['YearsCodePro'])

# ---- 1.1 Median ConvertedCompYearly by age group ----
median_by_age = df.groupby('Age')['ConvertedCompYearly'].median()
median_by_age = median_by_age.reindex([a for a in AGE_ORDER if a in median_by_age.index])
print(median_by_age)

plt.figure(figsize=(12, 6))
plt.plot(median_by_age.index, median_by_age.values, marker='o')
plt.title('Median Yearly Compensation by Age Group')
plt.xlabel('Age group')
plt.ylabel('Median ConvertedCompYearly (USD)')
plt.xticks(rotation=30, ha='right')
plt.grid(True)
plt.tight_layout()
plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
# ---- 1.2 Median ConvertedCompYearly for ages 25 to 45 ----
# The ranges covering 25-45 are '25-34 years old' and '35-44 years old'
subset = median_by_age.loc[['25-34 years old', '35-44 years old']]
print(subset)

plt.figure(figsize=(8, 5))
plt.plot(subset.index, subset.values, marker='o', color='seagreen')
plt.title('Median Yearly Compensation, Ages 25 to 45')
plt.xlabel('Age group')
plt.ylabel('Median ConvertedCompYearly (USD)')
plt.grid(True)
plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
# ---- 2.1 JobSatPoints_6 by experience level (YearsCodePro) ----
sat_by_exp = df[df['YearsCodePro_num'] <= 40].groupby('YearsCodePro_num')['JobSatPoints_6'].median()

plt.figure(figsize=(12, 6))
plt.plot(sat_by_exp.index, sat_by_exp.values, marker='o', markersize=4)
plt.title('Median Job Satisfaction (JobSatPoints_6) by Years of Professional Coding')
plt.xlabel('YearsCodePro (years)')
plt.ylabel('Median JobSatPoints_6')
plt.grid(True)
plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
# ---- 3.1 Median ConvertedCompYearly over experience (WorkExp) ----
comp_by_exp = df[df['WorkExp'] <= 40].groupby('WorkExp')['ConvertedCompYearly'].median()

plt.figure(figsize=(12, 6))
plt.plot(comp_by_exp.index, comp_by_exp.values, marker='o', markersize=4, color='purple')
plt.title('Median Yearly Compensation by Work Experience')
plt.xlabel('WorkExp (years)')
plt.ylabel('Median ConvertedCompYearly (USD)')
plt.grid(True)
plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
# ---- 3.2 JobSatPoints_6 across experience levels (WorkExp) ----
sat_by_work = df[df['WorkExp'] <= 40].groupby('WorkExp')['JobSatPoints_6'].median()

plt.figure(figsize=(12, 6))
plt.plot(sat_by_work.index, sat_by_work.values, marker='o', markersize=4, color='darkorange')
plt.title('Median Job Satisfaction (JobSatPoints_6) by Work Experience')
plt.xlabel('WorkExp (years)')
plt.ylabel('Median JobSatPoints_6')
plt.grid(True)
plt.show()
```
