## Identify and Analyze Duplicates


### Task 1: Identify Duplicate Rows
1. Count the number of duplicate rows in the dataset.
3. Display the first few duplicate rows to understand their structure.


```python
# Task 1: Identify duplicate rows
num_duplicates = df.duplicated().sum()
print("Number of fully duplicated rows:", num_duplicates)

# Show the first few duplicate rows (keep=False shows every copy, so you can compare them)
duplicate_rows = df[df.duplicated(keep=False)]
print("Rows involved in duplication (all copies):", len(duplicate_rows))
duplicate_rows.sort_values(by=list(df.columns[:3])).head(10)
```

### Task 2: Analysis of Duplicate Response Patterns
1. Identify duplicate response patterns based on selected columns such as MainBranch, Employment, and RemoteWork.
2. Clarify that these represent multiple respondents with identical answers rather than duplicate records. Analyse which other columns frequently share identical values within these response-pattern groups.
   


```python
# Task 2: Duplicate response patterns on MainBranch, Employment and RemoteWork
key_cols = ['MainBranch', 'Employment', 'RemoteWork']

pattern_dupes = df[df.duplicated(subset=key_cols, keep=False)]
print("Rows sharing a MainBranch/Employment/RemoteWork pattern with at least one other row:", len(pattern_dupes))

# Size of each response-pattern group
pattern_counts = (df.groupby(key_cols, dropna=False)
                    .size()
                    .reset_index(name='Respondents')
                    .sort_values('Respondents', ascending=False))
print(pattern_counts.head(10))

# NOTE: these are different respondents who happen to give the same answers,
# not duplicate records. ResponseId is still unique within each group:
if 'ResponseId' in df.columns:
    print("Unique ResponseIds among pattern rows:", pattern_dupes['ResponseId'].nunique())

# Which other columns most often share one identical value inside these groups?
other_cols = [c for c in ['Country', 'EdLevel', 'Age', 'DevType', 'YearsCodePro',
                          'OrgSize', 'Industry', 'LanguageHaveWorkedWith'] if c in df.columns]
shared = {}
for col in other_cols:
    # share of pattern groups (with 2+ members) in which this column has only one distinct value
    grp = pattern_dupes.groupby(key_cols, dropna=False)[col].nunique(dropna=False)
    shared[col] = round((grp == 1).mean() * 100, 1)
print("\n% of pattern groups where the column is identical for every member:")
print(pd.Series(shared).sort_values(ascending=False))
```

### Task 3: Visualize Shared Response Patterns
1. Create visualizations to show the distribution of shared response patterns across different categories.
2. Use bar charts or pie charts to represent the distribution of respondents who share identical values for MainBranch, Employment, and RemoteWork, grouped by Country and Employment.


```python
# Task 3: Visualize shared response patterns by Country and Employment
key_cols = ['MainBranch', 'Employment', 'RemoteWork']
shared_rows = df[df.duplicated(subset=key_cols, keep=False)]

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Top 10 countries among respondents who share a pattern
shared_rows['Country'].value_counts().head(10).plot(kind='bar', ax=axes[0], color='steelblue')
axes[0].set_title('Respondents with shared response patterns - Top 10 countries')
axes[0].set_xlabel('Country')
axes[0].set_ylabel('Number of respondents')
axes[0].tick_params(axis='x', rotation=75)

# Top employment categories among respondents who share a pattern
shared_rows['Employment'].value_counts().head(5).plot(kind='pie', ax=axes[1], autopct='%1.1f%%')
axes[1].set_title('Respondents with shared response patterns - Top 5 Employment types')
axes[1].set_ylabel('')

plt.tight_layout()
plt.show()
```
